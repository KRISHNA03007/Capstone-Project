```sql
-- Notice it is now just "CREATE LIVE TABLE" and reads from the table directly
CREATE LIVE TABLE silver_calendar (
  CONSTRAINT valid_date EXPECT (calendar_date IS NOT NULL) ON VIOLATION DROP ROW
)
COMMENT "Enriched calendar dimension"
AS SELECT
  calendar_date,
  YEAR(calendar_date) AS year,
  MONTH(calendar_date) AS month,
  DAY(calendar_date) AS day,
  QUARTER(calendar_date) AS quarter,
  DAYOFWEEK(calendar_date) AS day_of_week, 
  date_format(calendar_date, 'EEEE') AS day_name,
  date_format(calendar_date, 'MMMM') AS month_name,
  date_format(calendar_date, 'yyyy-MM') AS year_month,
  CASE 
    WHEN DAYOFWEEK(calendar_date) IN (1, 7) THEN TRUE 
    ELSE FALSE 
  END AS is_weekend,
  _ingestion_timestamp,
  _source_system
FROM mavenmarket.bronze.calendar;
```
