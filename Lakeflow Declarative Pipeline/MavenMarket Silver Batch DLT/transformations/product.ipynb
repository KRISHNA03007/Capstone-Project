```sql

CREATE LIVE TABLE silver_products (
  CONSTRAINT valid_product_id EXPECT (product_id IS NOT NULL) ON VIOLATION DROP ROW
)
COMMENT "Cleaned products master data"
AS SELECT
  _id,
  CAST(product_id AS INT) AS product_id,
  product_name,
  product_brand,
  product_category,
  product_sku,
  product_cost,
  CAST(product_retail_price AS DOUBLE) AS product_retail_price,
  CAST(product_weight AS DOUBLE) AS product_weight,
  low_fat,
  recyclable,
  _ingestion_timestamp,
  _source_system
FROM mavenmarket.bronze.products;
```
