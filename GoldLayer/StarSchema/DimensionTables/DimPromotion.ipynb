```sql
CREATE OR REFRESH MATERIALIZED VIEW mavenmarket.gold.dim_promotion
COMMENT 'Promotion Dimension Table'
AS
SELECT
    promotion_sk,
    promotion_id,
    TRIM(promotion_name) AS promotion_name,
    TRIM(promotion_type) AS promotion_type,
    CAST(discount_percent AS DECIMAL(5,2)) AS discount_percentage,
    start_date,
    end_date,
    TRIM(channel) AS marketing_channel,
    TRIM(media_type) AS media_type,
    __START_AT AS start_at_timestamp,
    __END_AT AS end_at_timestamp,
    __END_AT IS NULL AS is_current_flag
FROM mavenmarket.silver.silver_promotions
WHERE promotion_id IS NOT NULL; 
```
