```sql

CREATE OR REFRESH MATERIALIZED VIEW mavenmarket.gold.dim_date
COMMENT 'Date Dimension populated directly from silver_calendar source table'
AS
SELECT 
    CAST(date_format(calendar_date, 'yyyyMMdd') AS INT) AS date_sk,
    calendar_date,
    year AS year_number,
    month AS month_number,
    day AS day_of_month,
    quarter AS quarter_number,
    day_of_week AS day_of_week_number,
    TRIM(day_name) AS day_name,
    TRIM(month_name) AS month_name,
    TRIM(year_month) AS year_month_code,
    is_weekend AS is_weekend_flag
FROM mavenmarket.silver.silver_calendar
WHERE calendar_date IS NOT NULL;
```
