```sql

CREATE OR REFRESH MATERIALIZED VIEW mavenmarket.gold.dim_region
COMMENT 'Sales Region Dimension'
AS
SELECT 
    region_sk,
    CAST(region_id AS INT) AS region_id,
    TRIM(sales_district) AS sales_district_name,
    TRIM(sales_region) AS sales_region_name,
    __START_AT AS start_at_timestamp,
    __END_AT AS end_at_timestamp,
    __END_AT IS NULL AS is_current_flag
FROM mavenmarket.silver.silver_regions
WHERE region_id IS NOT NULL;
```
