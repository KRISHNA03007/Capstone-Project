```python
%sql
CREATE OR REPLACE MATERIALIZED VIEW mavenmarket.gold.dim_store
COMMENT 'Store Dimension Table with standardized business terms'
AS
SELECT 
    ABS(HASH(CAST(store_id AS STRING))) AS store_sk,
    store_id,
    ABS(HASH(CAST(region_id AS STRING))) AS region_sk,
    region_id,
    TRIM(store_type) AS store_type_description,
    TRIM(store_name) AS store_name,
    TRIM(store_street_address) AS street_address,
    TRIM(store_city) AS city_name,
    TRIM(store_state) AS state_province_code,
    TRIM(store_country) AS country_name,
    TRIM(store_phone) AS phone_number,
    TO_DATE(first_opened_date) AS first_opened_date,
    last_remodel_date,
    total_sqft AS total_area_sqft,
    grocery_sqft AS grocery_area_sqft
FROM mavenmarket.silver.silver_stores
WHERE store_id IS NOT NULL;
```
