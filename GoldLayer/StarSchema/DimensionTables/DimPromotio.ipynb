```python
%sql
CREATE OR REPLACE MATERIALIZED VIEW mavenmarket.gold.dim_promotion
COMMENT 'Promotion Dimension Table'
AS
SELECT 
    ABS(HASH(CAST(promotion_id AS STRING))) AS promotion_sk,
    promotion_id,
    TRIM(promotion_name) AS promotion_name,
    TRIM(promotion_type) AS promotion_type,
    CAST(discount_percent AS DECIMAL(5,2)) AS discount_percentage,
    start_date,
    end_date,
    TRIM(channel) AS marketing_channel,
    TRIM(media_type) AS media_type
FROM mavenmarket.silver.silver_promotions
WHERE promotion_id IS NOT NULL

UNION ALL

SELECT 
    ABS(HASH('0')) AS promotion_sk,
    0 AS promotion_id,
    'No Promotion' AS promotion_name,
    'None' AS promotion_type,
    CAST(0.00 AS DECIMAL(5,2)) AS discount_percentage,
    DATE '1900-01-01' AS start_date,
    DATE '2999-12-31' AS end_date,
    'None' AS marketing_channel,
    'None' AS media_type;
```
