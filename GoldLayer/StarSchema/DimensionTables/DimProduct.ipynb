```sql
CREATE OR REFRESH MATERIALIZED VIEW ${catalog}.${gold}.dim_product
COMMENT 'Product Dimension Table'
AS
SELECT 
    product_sk,
    CAST(product_id AS INT) AS product_id,
    TRIM(product_name) AS product_name,
    TRIM(product_brand) AS brand_name,
    TRIM(product_category) AS category_name,
    TRIM(product_sku) AS sku_code,
    CAST(product_cost AS DECIMAL(18,2)) AS unit_cost_amount,
    CAST(product_retail_price AS DECIMAL(18,2)) AS retail_price_amount,
    CAST(product_weight AS DECIMAL(10,2)) AS unit_weight_lbs,
    TRIM(low_fat) AS is_low_fat_flag,
    TRIM(recyclable) AS is_recyclable_flag,
    
    -- CDC / SCD Type 2 tracking columns directly from Silver
    __START_AT AS start_at_timestamp,
    __END_AT AS end_at_timestamp,
    __END_AT IS NULL AS is_current_flag
FROM ${catalog}.${silver}.silver_products
WHERE product_id IS NOT NULL;
```
